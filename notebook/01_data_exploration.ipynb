```python
import numpy as np
```

```python
data = np.genfromtxt(
    "../data/ETP_4_New_Data_Accidents.csv",
    delimiter = ",",
    dtype = None,
    encoding = "utf-8",
    names = True
)

print(data.dtype.names)
```

```python
print("Shape: ", data.shape)
print()
print(data[:5])
```

```python
for column in data.dtype.names:
    print(f"\n--- {column} ---")
    print(np.unique(data[column])[:20])
```

```python
categorical_columns = [
    "Accident_Location_A",
    "Accident_Location_A_Chainage_km_RoadSide",
    "Accident_Severity_C",
    "Causes_D",
    "Road_Feature_E",
    "Road_Condition_F",
    "Weather_Conditions_H",
    "Vehicle_Type_Involved_J_V1",
    "Vehicle_Type_Involved_J_V2"
]

for column in categorical_columns:
    values, counts = np.unique(data[column], return_counts=True)

    print(f"\n===== {column} =====")
    
    for value, count in zip(values, counts):
        print(f"{value}: {count}")
```

```python
for column in categorical_columns:
    values, counts = np.unique(data[column], return_counts=True)

    print(f"\n===== {column} =====")
    print("Unique values:", len(values))

    for value, count in zip(values, counts):
        print(f"{value!r} -> {count}")
```

```python
print("===== DATA QUALITY CHECK =====")

for column in data.dtype.names:
    values = data[column]

    print(
        f"{column:45} "
        f"Missing/negative: {np.sum(values == -1) if np.issubdtype(values.dtype, np.number) else 'N/A'}"
    )
```

```python
dates = data["Date"]

print("First date:", dates.min())
print("Last date:", dates.max())
```

```python
date_strings = data["Date"]

years = np.array([
    int(date.split("/")[-1])
    for date in date_strings
])

months = np.array([
    int(date.split("/")[1])
    for date in date_strings
])

print("Years:", np.unique(years))
print("Months:", np.unique(months))
```

```python
times = data["Time_of_Accident"]

hours = np.array([
    int(time.split(":")[0])
    for time in times
])

print("Hour range:", hours.min(), "to", hours.max())
print("Unique hours:", np.unique(hours))
```

```python
features = np.column_stack([
    years,
    months,
    hours,
    data["Day_of_Week"],
    data["Accident_Location_A_Chainage_km"],
    data["Accident_Location_A_Chainage_km_RoadSide"],
    data["Accident_Severity_C"]
])

print("Feature matrix shape:", features.shape)
print()
print(features[:10])
```

```python
print("===== FEATURE VALIDATION =====")

print("Shape:", features.shape)
print("NaN values:", np.isnan(features).sum())
print("Infinite values:", np.isinf(features).sum())
print("Minimum:", features.min(axis=0))
print("Maximum:", features.max(axis=0))
```

```python
print("\n===== COLUMN STATISTICS =====")

column_names = [
    "Year",
    "Month",
    "Hour",
    "Day",
    "Chainage",
    "RoadSide",
    "Severity"
]

for i, name in enumerate(column_names):
    print(
        f"{name:10} | "
        f"min={features[:, i].min():.2f} | "
        f"max={features[:, i].max():.2f} | "
        f"mean={features[:, i].mean():.2f}"
    )
```

```python
severity_values, severity_counts = np.unique(
    data["Accident_Severity_C"],
    return_counts=True
)

print("===== ACCIDENTS BY SEVERITY =====")

for severity, count in zip(severity_values, severity_counts):
    percentage = (count / len(data)) * 100

    print(
        f"Severity {severity}: "
        f"{count} accidents ({percentage:.2f}%)"
    )
```

```python
hour_values, hour_counts = np.unique(
    hours,
    return_counts=True
)

print("\n===== ACCIDENTS BY HOUR =====")

for hour, count in zip(hour_values, hour_counts):
    print(f"{hour:02d}:00 -> {count}")
```

```python
peak_hour_index = np.argmax(hour_counts)

print(
    "Peak accident hour:",
    hour_values[peak_hour_index]
)

print(
    "Accidents:",
    hour_counts[peak_hour_index]
)
```

```python
# Day: 06:00 to 17:59
# Night: 18:00 to 05:59

day_mask = (hours >= 6) & (hours < 18)
night_mask = ~day_mask

day_accidents = np.sum(day_mask)
night_accidents = np.sum(night_mask)

print("Day accidents:", day_accidents)
print("Night accidents:", night_accidents)

print("Day percentage:", round(day_accidents / len(data) * 100, 2))
print("Night percentage:", round(night_accidents / len(data) * 100, 2))
```

```python
chainage = data["Accident_Location_A_Chainage_km"]

print("Minimum chainage:", chainage.min())
print("Maximum chainage:", chainage.max())
print("Mean chainage:", chainage.mean())
print("Median chainage:", np.median(chainage))
print("Standard deviation:", chainage.std())
```

```python
segment_size = 10  # km

segments = np.floor(chainage / segment_size).astype(int)

segment_values, segment_counts = np.unique(
    segments,
    return_counts=True
)

print("Number of occupied segments:", len(segment_values))
print()

for segment, count in zip(segment_values[:20], segment_counts[:20]):
    start = segment * segment_size
    end = start + segment_size

    print(f"{start:3d}-{end:3d} km -> {count} accidents")
```

```python
top_indices = np.argsort(segment_counts)[-10:][::-1]

print("===== TOP 10 SEGMENTS =====")

for index in top_indices:
    segment = segment_values[index]
    count = segment_counts[index]

    print(
        f"{segment * 10:3d}-{segment * 10 + 10:3d} km "
        f"-> {count} accidents"
    )
```

```python
# Identify the segment with the most accidents
top_segment_index = np.argmax(segment_counts)

top_segment = segment_values[top_segment_index]

# Create mask for accidents inside that segment
top_segment_mask = segments == top_segment

top_segment_data = data[top_segment_mask]

print("===== TOP ACCIDENT SEGMENT =====")
print("Segment:", top_segment * 10, "-", top_segment * 10 + 10, "km")
print("Accidents:", np.sum(top_segment_mask))
print("Percentage of all accidents:",
      round(np.sum(top_segment_mask) / len(data) * 100, 2))
```

```python
severity_values, severity_counts = np.unique(
    top_segment_data["Accident_Severity_C"],
    return_counts=True
)

print("\n===== SEVERITY DISTRIBUTION =====")

for severity, count in zip(severity_values, severity_counts):
    percentage = count / len(top_segment_data) * 100

    print(
        f"Severity {severity}: "
        f"{count} ({percentage:.2f}%)"
    )
```

```python
overall_severity_values, overall_severity_counts = np.unique(
    data["Accident_Severity_C"],
    return_counts=True
)

print("\n===== TOP SEGMENT vs OVERALL =====")

for severity in overall_severity_values:

    segment_count = np.sum(
        top_segment_data["Accident_Severity_C"] == severity
    )

    overall_count = np.sum(
        data["Accident_Severity_C"] == severity
    )

    segment_pct = segment_count / len(top_segment_data) * 100
    overall_pct = overall_count / len(data) * 100

    print(
        f"Severity {severity}: "
        f"Segment={segment_pct:.2f}% | "
        f"Overall={overall_pct:.2f}%"
    )
```

```python
severity_difference = []

for severity in overall_severity_values:

    segment_count = np.sum(
        top_segment_data["Accident_Severity_C"] == severity
    )

    overall_count = np.sum(
        data["Accident_Severity_C"] == severity
    )

    segment_pct = segment_count / len(top_segment_data) * 100
    overall_pct = overall_count / len(data) * 100

    difference = segment_pct - overall_pct

    severity_difference.append(difference)

    print(
        f"Severity {severity}: "
        f"{difference:+.2f} percentage points"
    )
```

```python
largest_difference_index = np.argmax(
    np.abs(severity_difference)
)

print(
    "\nLargest distribution deviation:",
    overall_severity_values[largest_difference_index]
)

print(
    "Difference:",
    severity_difference[largest_difference_index],
    "percentage points"
)
```

```python
segment_summary = np.column_stack([
    segment_values,
    segment_counts,
    (segment_counts / len(data)) * 100
])

# Calculate mean severity for each segment
mean_severity = np.array([
    data["Accident_Severity_C"][segments == segment].mean()
    for segment in segment_values
])

segment_summary = np.column_stack([
    segment_summary,
    mean_severity
])

print("Shape:", segment_summary.shape)
print()
print("Segment | Accidents | % Total | Mean Severity")

for row in segment_summary[:10]:
    print(
        f"{int(row[0])*10:3d}-{int(row[0])*10+10:3d} km | "
        f"{int(row[1]):4d} | "
        f"{row[2]:6.2f}% | "
        f"{row[3]:.2f}"
    )
```

```python
# Sort segments by accident count
ranked_indices = np.argsort(segment_summary[:, 1])[::-1]

print("===== TOP 10 BY ACCIDENT COUNT =====")

for index in ranked_indices[:10]:
    row = segment_summary[index]

    print(
        f"{int(row[0])*10:3d}-{int(row[0])*10+10:3d} km | "
        f"{int(row[1]):4d} accidents | "
        f"{row[2]:.2f}% | "
        f"Mean severity: {row[3]:.2f}"
    )
```
